## Create Directories with Top Passages for ECCO-TCP Search Engines (both Vector and Hathi-Simulator)

### Overview

Much like the previous notebook, this creates directories filled with top-passage `.txt` files of top passages for the two search engines built on ECCO-TCP.

### 1. Libraries

```python
import os
import shutil
import pandas as pd
```

### 2. Configuration

```python
# 1. Define the paths
# The working directory where the CSV lives
base_project_dir = "/Users/sosadetz/Documents/01-Work/01. Ongoing Projects and Scrivener/03. Benchmark Development/aes-benchmark-supplement"

# The specific CSV file
csv_file_path = os.path.join(base_project_dir, "4_results/ht_eccotcp_principle_psgs.csv")

# The source directory where the original text files live
source_directory = "/Users/sosadetz/Documents/01-Work/02. DH and Search/ecco_tcp_standardized_v2/complete_ecco_tcp_passages"

# The target directory where you want the copies to go
target_directory = os.path.join(base_project_dir, "2_data", "ht_eccotcp_passages", "principle")

# 2. Configuration options
limit = 100
filename_column = "chunk_id"
```

### 3. Create and populate new directory, copying ECCO-TCP passage files

```python
# Create the target directory (and parent folders) if they don't exist
os.makedirs(target_directory, exist_ok=True)
print(f"Target directory verified: {target_directory}")

try:
    # Load the CSV
    df = pd.read_csv(csv_file_path)
    
    # Select the first 100 rows
    top_100 = df.head(limit)
    
    print(f"Found {len(df)} rows in CSV. Processing the top {len(top_100)}...")
    
    success_count = 0
    missing_count = 0
    
    for index, row in top_100.iterrows():
        # Get the filename from the column
        filename = str(row[filename_column]).strip()
        
        # Construct full file paths
        src_path = os.path.join(source_directory, filename)
        dst_path = os.path.join(target_directory, filename)
        
        # Check if the source file actually exists
        if os.path.exists(src_path):
            # Copy the file (copy2 preserves metadata)
            shutil.copy2(src_path, dst_path)
            success_count += 1
        else:
            print(f"WARNING: Source file not found: {filename}")
            missing_count += 1

    print(f"\n--- Operation Complete ---")
    print(f"Successfully copied: {success_count}")
    if missing_count > 0:
        print(f"Missing files: {missing_count}")
        
except FileNotFoundError:
    print(f"Error: Could not find the CSV file at {csv_file_path}")
except KeyError:
    print(f"Error: The column '{filename_column}' was not found in the CSV.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
