## Extract Passages from Full ECCO Cluster Search Engine Results (i.e. the primary search platform)

### Overview¶

This notebook creates a directory of passage-level `.txt` files. 
Subsequently, these passages will be matched with corresponding ECCO-TCP passages and deduplicated with results from the other search engines.

### 1. Import Libraries

```python
import csv
import os
```

### 2. Configuration

```python
input_csv_filename = 'principle_cluster_passages_full.csv'
output_directory = os.path.join('2_data', 'cluster_passages', 'principle_cluster_psgs')
files_to_create = 100
```

### 3. Create and populate new directory

```python
# 1. Create the target directory structure (and parent folders) if they don't exist
os.makedirs(output_directory, exist_ok=True)
print(f"Target directory verified: {output_directory}")

created_count = 0

try:
    # 2. Open the CSV file
    with open(input_csv_filename, 'r', encoding='utf-8') as f:
        # DictReader automatically uses the header row (Row 1) as keys
        reader = csv.DictReader(f)
        
        print(f"Processing the first {files_to_create} rows...")
        
        for row in reader:
            # Stop after creating the desired number of files
            if created_count >= files_to_create:
                break
            
            # 3. Extract the specific columns
            # .strip() removes accidental whitespace around the ID or text
            doc_id = row.get('document_id', '').strip()
            content = row.get('highlighted_passages', '')
            
            if doc_id:
                # 4. Construct the file path
                # Ensure the filename ends in .txt
                filename = f"{doc_id}.txt"
                file_path = os.path.join(output_directory, filename)
                
                # 5. Write the content to the TXT file
                with open(file_path, 'w', encoding='utf-8') as out_file:
                    out_file.write(content)
                
                created_count += 1
            else:
                print(f"Skipped a row because 'document_id' was missing.")

    print(f"\nSuccess! Created {created_count} text files in '{output_directory}'.")

except FileNotFoundError:
    print(f"Error: The file '{input_csv_filename}' was not found in the current directory.")
except KeyError as e:
    print(f"Error: Could not find column {e} in the CSV. Please check your headers.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
